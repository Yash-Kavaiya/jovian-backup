```python
import requests
from bs4 import BeautifulSoup
import openpyxl

# Set the search parameters
state = 'Maharashtra'
district = 'Nagpur'
page_size = 10
total_entries = 1254

# Initialize the Excel workbook and worksheet
workbook = openpyxl.Workbook()
worksheet = workbook.active
worksheet.append(['NGO Name', 'Address', 'City', 'State', 'Pincode', 'Owner Name', 'Email', 'Phone'])

# Loop through each page of the search results
for page in range(1, total_entries // page_size + 2):
    url = f'https://ngodarpan.gov.in/index.php/home/statewise_ngo/228/35/{page}?&per_page={page_size}'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find the table that contains the NGO information
    table = soup.find('table', {'class': 'table'})

    # Find all the rows in the table
    rows = table.find_all('tr')

    # Loop through each row and extract the NGO details
    for row in rows[1:]:
        columns = row.find_all('td')
        ngo_name = columns[1].text.strip()
        address = columns[2].text.strip()
        city = columns[3].text.strip()
        state = columns[4].text.strip()
        #pincode = columns[5].text.strip()

        # Find the hyperlink to the NGO details page
        # Find the hyperlink to the NGO details page
        link = columns[1].find('a')
        if link is None:
            continue
        details_url = f"https://ngodarpan.gov.in{link['href']}"
        details_response = requests.get(details_url)
details_soup = BeautifulSoup(details_response.text, 'html.parser')


        # Find the owner's name, email, and phone number
        owner_name = ''
        email = ''
        phone = ''
        table = details_soup.find('table', {'class': 'table'})
        if table is not None:
            rows = table.find_all('tr')
            for row in rows:
                header = row.find('th')
                if header is not None and 'Name' in header.text:
                    owner_name = row.find('td').text.strip()
                elif header is not None and 'Email' in header.text:
                    email = row.find('td').text.strip()
                elif header is not None and 'Mobile No.' in header.text:
                    phone = row.find('td').text.strip()

        worksheet.append([ngo_name, address, city, state, pincode, owner_name, email, phone])
        print(f'Added {ngo_name} to the worksheet')

# Save the Excel workbook
workbook.save('ngos.xlsx')
```
