```python
base_url = 'https://ngodarpan.gov.in/index.php/home/statewise_ngo/'
state_range = range(1, 301)  # Pages 1 to 300
district_range = range(1, 501)  # Pages 1 to 500
per_page = 100  # Results per page

urls = []

for state in state_range:
    for district in district_range:
        url = f"{base_url}{state}/{district}/1?per_page={per_page}"
        urls.append(url)
```

```python

for url in urls:
    response = requests.get(url)
    # Process the response as needed
    soup = BeautifulSoup(response.text, 'html.parser')
    table = soup.find('table', {'class': 'table'})
```
