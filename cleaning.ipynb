```python
df = df.drop(columns=['number', 'icao24', 'registration','typecode','firstseen','lastseen','latitude_1','longitude_1'
,'altitude_1','latitude_2','longitude_2','altitude_2'], axis=1)
```

```python
df.shape
```

```python
df
```

```python
airports=pd.read_csv("/content/drive/Shareddrives/Data Anaytics with Python :NPTEL/airports.csv")
```

```python
airports
```

```python
df = df.merge(airports[['gps_code','iso_region','type']], left_on='origin', right_on='gps_code', how='left')
df = df.rename(columns={'iso_region':'orig_region','type':'orig_type'})
df = df.merge(airports[['gps_code','iso_region','type']], left_on='destination', right_on='gps_code', how='left')
df = df.rename(columns={'iso_region':'dest_region','type':'dest_type'})
```

```python
df
```

```python
df['dest_region'].fillna('unknown', inplace=True)
```

```python
df = df[df['dest_region'].str.startswith('US-')]
```

```python
import numpy as np
```

```python


# 6. Use orig_region to:
# If origin_region = US-XX then create new column flight_type with value Domestic
# If origin_region ≠ US-XX then fill flight_type column with value International
df['flight_type'] = np.where(df['orig_region'].str.startswith('US-'), 'Domestic', 'International')

# 7. Remove time from day column and keep only date YYYY-MM-DD
df['day'] = df['day'].str[:10]

# 8. Save file as .csv.gz
df.to_csv("main_dataset_processed.csv.gz", compression='gzip', index=False)
```
