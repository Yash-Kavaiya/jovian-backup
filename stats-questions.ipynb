1. Suppose that a year has 365 days and that birthdays are equally distributed over the year. In our class, we have 250 students, what's the probability that at least two students have the same birthday? (Enter your values to 3 significant figures): 

This is a classic problem in probability theory known as the "Birthday Problem". To solve it, we can use the formula for the probability of an event's complement, which is 1 minus the probability of the event itself.

Let's start by calculating the probability that no two students have the same birthday. The first student can have any birthday, so the probability is 1. The second student must have a different birthday, which has a probability of 364/365 (since there are 364 days left in the year). Similarly, the third student must have a birthday different from the first two, which has a probability of 363/365. Continuing this pattern, the probability that no two students have the same birthday is:

$$ 1 * {364}/{365} * 363/365 * ... * (365-249+1)/365 $$

We can simplify this expression by noting that the product is equal to the ratio of factorials (i.e., 365!/[(365-250)!365^250]), so we can compute it numerically as:



This gives us p_no_match = 0.2937 (to 4 decimal places).

Therefore, the probability that at least two students have the same birthday is:

p_match = 1 - p_no_match
p_match

This gives us p_match = 0.7063 (to 4 decimal places), or approximately 0.706. Therefore, the probability that at least two students have the same birthday is about 70.6%

```python
# define the number of students and days in a year
n_students = 250
n_days = 365

# calculate the probability of no two students having the same birthday
prob_no_match = 1
for i in range(n_students):
    prob_no_match *= (n_days - i) / n_days

# calculate the probability of at least two students having the same birthday
prob_match = 1 - prob_no_match

# print the result
print(f"The probability of at least two students having the same birthday is {prob_match:.3f}")
```

## 2. You are taking a 4-question multiple choice test. If each question has three choices and you randomly guess on each question. 
Answer the following questions (enter your values to 3 significant figures): 


2.1. What's the probability that you will get exactly 1 question correctly? (5 mark) 

```python
import math

def calculate_probability(k):
    n = 4  # number of questions
    # number of correct answers
    p = 1/3  # probability of guessing the correct answer
    
    # calculate the binomial coefficient (n choose k)
    binomial_coefficient = math.comb(n, k)
    
    # calculate the probability of getting exactly 1 question correct
    probability = binomial_coefficient * p**k * (1-p)**(n-k)
    
    return probability
```

```python
probability = calculate_probability(1)
print(f"The probability of getting exactly 1 question correct is {probability:.3f}")
```

2.2. What's the probability that you will get exactly 2 questions correctly (5 mark) 

```python
probability = calculate_probability(2)
print(f"The probability of getting exactly 2 questions correct is {probability:.3f}")
```

2.3. What's the expected number of questions (expected value, does not have to be integers) you will answer correctly? (5 marks) 

```python
def calculate_expected_value():
    n = 4  # number of questions
    p = 1/3  # probability of guessing the correct answer
    
    # calculate the expected value
    expected_value = n * p
    
    return expected_value
expected_value = calculate_expected_value()
print(f"The expected number of questions answered correctly is {expected_value:.3f}")
```

# Question 4 (5 marks)  Here are 21 numbers in increasing order: 
0, 4, 14, 21, 28, 36, 37, 40, 41, 43, 47, 54, 63, 67, 69, 90, 90, 93, 94, 96, 99 
What's the mean, median, and mode, respectively? 


```python
import statistics

# Define the list of numbers
numbers = [0, 4, 14, 21, 28, 36, 37, 40, 41, 43, 47, 54, 63, 67, 69, 90, 90, 93, 94, 96, 99]

# Calculate the mean
mean = sum(numbers) / len(numbers)
median = statistics.median(numbers)

# Calculate the mode
mode = statistics.mode(numbers)

# Print the results
print("Mean: {:.3f}".format(mean))
print("Median: {}".format(median))
print("Mode: {}".format(mode))
```

# Question 5
Will students who perform very badly on Test 1 tend to perform less badly on Test 2? (4 marks) 
- a. Yes 
- b. No

What's the statistical phenomenon behind this? (6 marks) 
- a. Simpson paradox 
- b. Berkson paradox 
- c. Regression to the mean 
- d. Maximum likelihood 

The answer to the first question is b. No. Generally, students who perform very badly on one test tend to perform badly on subsequent tests as well.

The statistical phenomenon behind this is c. Regression to the mean. This phenomenon occurs when an extreme score on one test is likely to be followed by a less extreme score on a subsequent test, due to the effects of chance or measurement error. In other words, extreme scores are likely to be less extreme on a subsequent test simply because they were extreme on the first test. However, this does not mean that the subsequent scores will be good, only that they will tend to be less extreme than the initial scores.

Simpson's paradox (a) is a phenomenon in statistics where a trend appears in different groups of data but disappears or reverses when these groups are combined. Berkson's paradox (b) is a phenomenon in probability theory that arises when two factors that are negatively correlated are used to predict a third factor, leading to an apparent positive correlation between the two factors. Maximum likelihood (d) is a statistical method used for estimating the parameters of a statistical model based on observed data.
