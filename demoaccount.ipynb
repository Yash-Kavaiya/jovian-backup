```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from scipy.stats import loguniform
from scipy.stats import uniform
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso
from sklearn.linear_model import LassoCV
from sklearn.linear_model import RidgeCV
from sklearn.linear_model import Ridge
from sklearn.linear_model import SGDRegressor
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_absolute_percentage_error
```

```python
from sklearn.model_selection import cross_validate
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import ShuffleSplit
from sklearn.model_selection import validation_curve
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.pipeline import Pipeline
```

```python
np.random.seed(306)
```

```python
cv=ShuffleSplit(n_splits=10,test_size=0.2,random_state=42)
```

```python
X,y=fetch_california_housing(as_frame=True,return_X_y=True)
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train,y_test=train_test_split(X,y,random_state=42)
```

```python
lrpip=Pipeline([("fs",StandardScaler()),("lr",LinearRegression())])
```

```python
lrcv=cross_validate(lrpip,X_train,y_train,cv=cv,scoring='neg_mean_absolute_error',return_estimator=True,return_train_score=True)
```

```python
pd.DataFrame(lrcv)
```

```python
sgdpip=Pipeline([("fs",StandardScaler()),("lr",SGDRegressor(max_iter=np.ceil(1e6/X_train.shape[0]),early_stopping=True,eta0=1e-4,learning_rate='constant',tol=1e-5,validation_fraction=0.1,n_iter_no_change=5,random_state=42))])
```

```python
sgdcv=cross_validate(sgdpip,X_train,y_train,cv=cv,scoring='neg_mean_absolute_error',return_estimator=True,return_train_score=True)
```

```python
pd.DataFrame(sgdcv)
```
